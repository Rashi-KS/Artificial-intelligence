```python
def alpha_beta(node, depth, alpha, beta, maximizing, pruned):
    if isinstance(node, int):  # leaf
        return node

    if maximizing:
        value = float('-inf')
        for child in node:
            # Before evaluating child, record current alpha for detecting pruning later
            old_alpha = alpha

            eval_val = alpha_beta(child, depth - 1, alpha, beta, False, pruned)
            value = max(value, eval_val)
            alpha = max(alpha, value)

            # Pruning occurs
            if alpha >= beta:
                # Add ALL remaining leaf values in this child as pruned
                collect_pruned(child, pruned)
                break

        return value

    else:  # MIN node
        value = float('inf')
        for child in node:
            old_beta = beta

            eval_val = alpha_beta(child, depth - 1, alpha, beta, True, pruned)
            value = min(value, eval_val)
            beta = min(beta, value)

            # Pruning occurs
            if beta <= alpha:
                collect_pruned(child, pruned)
                break

        return value


# Helper: collect ALL leaves inside a pruned subtree
def collect_pruned(node, pruned):
    if isinstance(node, int):
        pruned.append(node)
    else:
        for child in node:
            collect_pruned(child, pruned)


# ---------------------------
# YOUR EXACT TREE
# ---------------------------
tree = [
    [               # Left MIN
        [10, 9],    # MAX
        [14, 18]    # MAX  <-- 18 gets pruned
    ],
    [               # Right MIN
        [5, 4],     # MAX
        [50, 3]     # MAX  <-- both 50 and 3 pruned
    ]
]


# Run alpha-beta
alpha = float('-inf')
beta = float('inf')
pruned_nodes = []

print("Initial alpha =", alpha)
print("Initial beta  =", beta)

root_value = alpha_beta(tree, 3, alpha, beta, True, pruned_nodes)

print("\nPruned values:", pruned_nodes)
print("\nFinal Root Node Value =", root_value)
```
